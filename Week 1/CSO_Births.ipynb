```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
#Projected Births#
births = pd.read_csv("https://raw.githubusercontent.com/DBrennan88/Programming-for-Data-Analytics/refs/heads/main/CSO.csv")
births.head()
print("Highest:", births["VALUE"].max())
print("Lowest:", births["VALUE"].min()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(births["Year"], births["VALUE"], marker="o")
plt.title("Projected annual births in Ireland")
plt.xlabel("Year")
plt.ylabel("Births per year")
plt.grid(True)
plt.show()
```
